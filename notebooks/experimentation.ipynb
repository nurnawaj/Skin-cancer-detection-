```python
import os, sys
sys.path.append(os.path.abspath(".."))

from config import config as cfg
import pandas as pd

metadata = pd.read_csv(cfg.METADATA_PATH)
metadata.head()
```

```python
# Class distribution
metadata[cfg.LABEL_COLUMN].value_counts().plot(kind="bar", figsize=(8, 4), color="#3B82F6")
```

```python
# Preview a few sample images with their labels
import matplotlib.pyplot as plt
from PIL import Image

sample = metadata.sample(6, random_state=cfg.RANDOM_SEED)
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, (_, row) in zip(axes.flatten(), sample.iterrows()):
    img_path = os.path.join(cfg.IMAGES_PATH, row[cfg.IMAGE_ID_COLUMN] + cfg.IMAGE_EXTENSION)
    if os.path.exists(img_path):
        ax.imshow(Image.open(img_path))
    ax.set_title(row[cfg.LABEL_COLUMN])
    ax.axis("off")
plt.tight_layout()
```
